## (1) Dataset principal de inmuebles
- Enlace: https://www.kaggle.com/datasets/toriqulstu/worlds-real-estate-data147k
- Información: anuncios de propiedades a nivel global (precio en USD, país, ubicación, habitaciones, baños, área, año de construcción, etc.).

## (2) PIB por país (PPP)
- Enlace: https://www.kaggle.com/code/mauricioasperti/crime-index-study-and-prediction-by-country
- Información: PIB por paridad de poder adquisitivo por país, útil para contexto económico.

## (3) Desigualdad de ingresos
- Enlace: https://www.kaggle.com/code/mauricioasperti/crime-index-study-and-prediction-by-country
- Información: índice Gini por país y año de referencia, útil para medir desigualdad social.

## (4) Población por país
- Enlace: https://www.kaggle.com/code/mauricioasperti/crime-index-study-and-prediction-by-country
- Información: población 2018 y 2019, variación porcentual y región ONU.

## (5) Índice de crimen
- Enlace: https://www.kaggle.com/code/mauricioasperti/crime-index-study-and-prediction-by-country
- Información: índice de crimen y de seguridad por país.

## (6) Tasa de desempleo
- Enlace: https://www.kaggle.com/code/mauricioasperti/crime-index-study-and-prediction-by-country
- Información: desempleo actual, previo y fecha de referencia por país.

## (7) Latitud y longitud de países
- Enlace: https://www.kaggle.com/code/mauricioasperti/crime-index-study-and-prediction-by-country
- Información: coordenadas geográficas y código de país para análisis y mapas.

## (8) Códigos ISO para visualización
- Enlace: plotly_countries_and_codes.csv
- Información: mapeo de nombre de país a código ISO-3 para mapas y gráficos de Plotly.

## (9) Costo de vida
- Enlace: cost_of_living.csv
- Información: índices comparativos de costo de vida por país/ciudad.

## (10) Área por país
- Enlace: https://www.kaggle.com/code/mauricioasperti/crime-index-study-and-prediction-by-country
- Información: superficie total por país para contexto territorial.

## (11) Densidad de población
- Enlace: Densidad_poblacion.csv
- Información: habitantes por km2 para medir concentración demográfica.

## (12) Índice de educación
- Enlace: https://www.kaggle.com/code/mauricioasperti/crime-index-study-and-prediction-by-country
- Información: indicador comparativo del nivel educativo por país.
